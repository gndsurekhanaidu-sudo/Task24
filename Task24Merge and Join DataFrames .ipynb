```python
import pandas as pd

customers = pd.DataFrame({
    'Customer_ID': [1, 2, 3],
    'Name': ['Ravi', 'Sita', 'Ram']
})

orders = pd.DataFrame({
    'Order_ID': [101, 102, 103],
    'Customer_ID': [1, 2, 4],
    'Amount': [500, 700, 300]
})

print("Customers:")
print(customers)

print("Orders:")
print(orders)
```

```python
inner = pd.merge(customers, orders,
                 on='Customer_ID',
                 how='inner')

print(inner)
```

```python
left = pd.merge(customers, orders,
                on='Customer_ID',
                how='left')

print(left)
```

```python
right = pd.merge(customers, orders,
                 on='Customer_ID',
                 how='right')

print(right)
```

```python
outer = pd.merge(customers, orders,
                 on='Customer_ID',
                 how='outer')

print(outer)
```

```python
result = pd.merge(customers, orders,
                  on='Customer_ID',
                  how='outer',
                  indicator=True)

print(result)
```

```python
final_data = pd.merge(customers, orders,
                      on='Customer_ID',
                      how='outer')

print("Final Combined Dataset:")
print(final_data)

final_data.to_csv("final_dataset.csv", index=False)
```
